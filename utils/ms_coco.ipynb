```python
from pycocotools.coco import COCO
import requests
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
annFile = 'annotations/captions_val2017.json'
if not os.path.exists(annFile):
    os.system("wget http://images.cocodataset.org/annotations/annotations_trainval2017.zip")
    os.system("unzip annotations_trainval2017.zip")
```

```python
# Initialize COCO API
coco = COCO(annFile)

# Get one image
img_id = coco.getImgIds()[0]
img_info = coco.loadImgs(img_id)[0]
```

```python
# Download image
img_path = f"http://images.cocodataset.org/val2017/{img_info['file_name']}"
img = Image.open(requests.get(img_path, stream=True).raw)

# Show image
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
# Print captions
ann_ids = coco.getAnnIds(imgIds=img_id)
anns = coco.loadAnns(ann_ids)
for ann in anns:
    print("-", ann['caption'])
```
